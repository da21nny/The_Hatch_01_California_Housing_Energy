##### Configuración e Importación de Librerías

```python
# Importar librerias de Data Science
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuracion de los estilos para graficos
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

# Carga del Dataset housing
df_housing = pd.read_csv("docs/housing.csv")
```

##### Exploración de Características (EDA Inicial)

```python
# Inspeccion rapidad de la estructura y ipo de datos
df_housing.info()
df_housing.describe().round(2)

# Visualizar distribuciones de caracteristicas individuales
for columna in df_housing.select_dtypes(include=[np.number]).columns:
    plt.figure()
    sns.histplot(df_housing[columna], kde=True, color="skyblue")
    plt.title(f"Distribucion de {columna}")
    plt.show

# Analizar la variable categorica clave: 'Ocean Proximity' o similar
sns.countplot(data=df_housing, x='ocean_proximity')
plt.title("Distribucion de la Cercania al Oceano")
plt.show()
```
